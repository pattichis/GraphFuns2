## Usar ciclos for
Digamos que queremos contar hasta 100.

¡Tendríamos que repetir estas instrucciones 100 veces!

Los ciclos <tt>for</tt> nos permiten usar una sola instrucción con la función <tt>print()</tt>.

Para usar un <tt>for</tt>, tenemos que definir el número con que empezamos y cuándo dejamos de contar.

Volvamos a hacer el código para contar 1, 2, 3:
```
startNumber = 1
stopNumber  = 3 + 1
```
Aquí, <tt>stopNumber = 3 + 1</tt> quiere decir que debemos dejar de contar para no pasarnos de 3.

Para contar 1, 2, 3, usamos el ciclo <tt>for</tt> como se muestra abajo:
```
for i in range(startNumber, stopNumber):
  print(i)
```
Aquí, <tt>range(startNumber, stopNumber)</tt> genera los números 1, 2, 3.

Empieza con 1 (=<tt>startNumber</tt>), luego 2, 3, pero como <tt>stopNumber=4</tt>, no genera el 4.

Luego la línea <tt>i in range(startNumber, stopNumber)</tt> copia estos números en <tt>i</tt>, uno por uno.

Después de copiar cada número, ejecuta <tt>print(i)</tt> para imprimirlo.

## Actividad: contar hasta otros valores

¡Cambia el código para contar hasta 100!







````python
startNumber = 1
stopNumber  = 3+1
for i in range(startNumber, stopNumber):
  print(i)
````

# Crear listas con ciclos
Podemos guardar los resultados en listas.<br>
Para hacerlo, empezamos con listas vacías fuera del ciclo:
```
x_vals = []
```
Luego usamos <tt>append</tt> dentro del ciclo:
```
x_vals   = []
for x in range(startNumber, stopNumber, step):
  # aquí va el código del ciclo ...
  x_vals.append(x)
```
De la misma manera, podemos ir guardando las evaluaciones de la función y los resultados
en una lista.<br>
Mira el código de abajo.

## Actividad:
Estudia el código que se te dio.
1. Imprime las tres listas que se generaron.
2. ¿Coinciden con el orden del resultado? Explica.





````python
step = 1
startNumber = -2
stopNumber  = 4 + step

x_vals   = []
y_vals   = []
str_vals = []
for x in range(startNumber, stopNumber, step):
  y = 0.05*(2)**x

  str_val = "y=0.05*(2)**"+str(x)

  print("x=", x, str_val, x, " y=", y)

  x_vals.append(x)
  y_vals.append(y)
  str_vals.append(str_val)
````

# Crear tablas con programación orientada a objetos
Podemos crear una tabla bonita con los resultados.<br>
Aquí creamos un objeto de tabla usando:
```
my_table = table()
```
Luego podemos guardar las etiquetas de las columnas usando una lista de cadenas de texto:
```
my_table.column_labels = ["x", "y=0.05*(2)**x", "y"]
```
Los valores se pueden guardar usando otra lista:
```
my_table.data_values   = [x_vals, str_vals, y_vals]
```
Luego podemos imprimir la tabla usando:
```
my_table.plotTable()
```

````python
# Crea una instancia de un objeto de tabla
my_table = table()

# Etiquetas
my_table.column_labels = ["x", "y=0.05*(2)**x", "y"]

# Datos
my_table.data_values   = [x_vals, str_vals, y_vals]

print(my_table.data_values)

# Dibuja la tabla
my_table.plotTable()
````
