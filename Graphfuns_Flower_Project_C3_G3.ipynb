# Make a beautiful flower

1. Make the first pedal look nice by reducing the domain.
2. Create the other pedals. Have fun coloring your pedals.
3. After you get the pedals, set hide_axes=True in graph_funs().
4. Remove plot.plot_funs() function calls to plot one pedal at a time only.
5. Save each pedal into a different picture.

Here is a list of colors:
   "blue", "green", "red", "cyan", "magenta", "yellow", "black", "white","olive","indigo".

```python
# MODIFY to add multiple transformations

# Define x as a symbolic variable
x = sp.symbols('x')

# Define exponential function
f1 = 0.25*(2)**(x)

# Create a graph object with equal_axes)
plot = graph_funs(equal_axes=True, hide_axes=True)

# Setup the grid
# Determine the range for X based on reflections and translations.
# Determine the range for Y based on reflections and translations (look at maximum values).
plot.set_XY(minX=-10, maxX=+10, minY=-10, maxY=+10)

# Add to the list of functions:
domain1 = [0.33, 4, 100]

# Plot fun
plot.add_fun(f1, domain1, 'f1(x)='+str(f1), 'light blue')  # Add the original function.
plot.plot_funs(plot_title="Original function f1 (light blue)", img_name="frame1.png")



# Add the inverse function.
domain3 = domain1
f3 = plot.add_inverse_fun(f1, domain3, 'inverse of f1', 'light blue')
plot.plot_funs(plot_title="f1, f2, inverse f1", img_name="frame2.png")

# Flip the inverse function:
f4 = -f3
domain4 = domain1
plot.add_fun(f4, domain4, 'f4', 'light blue')
plot.plot_funs(plot_title="f1, f2, f3, f4", img_name="frame3.png")

# Reflect f4:
f5 = f4.subs(x, -x)
domain5 = [-4, -0.33, 100]
plot.add_fun(f5, domain5, 'f5', 'light blue')
plot.plot_funs(plot_title="f1, f2, f3, f4, f5", img_name="frame4.png")

# Reflect f5:
f6 = -f5
domain6 = domain5
plot.add_fun(f6, domain6, 'f6', 'light blue')
plot.plot_funs(plot_title="f1, f2, f3, f4, f5, f6", img_name="frame5.png")



f8 = 0.25*(2)**(-x)
domain8 = [-4, -0.33, 100]
plot.add_fun(f8, domain8, 'f8', 'light blue')
plot.plot_funs(plot_title="f1-f8", img_name="frame6.png")

f9 = -0.25*(2)**(x)
domain9 = [0.33, 4, 100]
plot.add_fun(f9, domain9, 'f9', 'light blue')
plot.plot_funs(plot_title="f1-f9", img_name="frame7.png")


f10 = -0.25*(2)**(-x)
domain10 = [-4, -.33, 100]
plot.add_fun(f10, domain10, 'f10', 'light blue')
plot.plot_funs(plot_title="f1-f10", img_name="frame8.png")


#plot.plot_funs(plot_title="f1, f2, inverse f1", img_name="frame1.png")
#plot.plot_funs(plot_title="f1-f8", img_name="frame2.png")
#plot.plot_funs(plot_title="f1-f9", img_name="frame4.png")
```
